```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("ahmedmohamed2003/cafe-sales-dirty-data-for-cleaning-training")

print("Path to dataset files:", path)
```

```python
import pandas as pd
import numpy as np
import os

file_name = "dirty_cafe_sales.csv"
file_path = os.path.join(path, file_name)

df = pd.read_csv(file_path)
```

```python
df['Total Spent'] = df['Total Spent'].replace('ERROR', 0)
df['Total Spent'] = pd.to_numeric(df['Total Spent'], errors='coerce')
```

```python
from csv import Error
ERROR = "ERROR"
(df['Total Spent'] == ERROR).sum()
```

```python
for i in df['Total Spent']:
  if i == ERROR:
    i = 0
```
